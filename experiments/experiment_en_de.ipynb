```python
from cs336_basics import BPETokenizer
from pathlib import Path
import time
import logging
import os
import random

logger = logging.getLogger(__name__)
logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s %(levelname)s %(name)s: %(message)s",
    datefmt="%Y-%m-%d %H:%M:%S",
)

current_dir = Path(os.getcwd())
random.seed(42)  # For reproducibility
```

```python
tiny_vocab_path = str((current_dir / "vocab_TinyStories.json").resolve())
owt_vocab_path = str((current_dir / "vocab_owt.json").resolve())
tiny_merges_path = str((current_dir / "merges_TinyStories.json").resolve())
owt_merges_path = str((current_dir / "merges_owt.json").resolve())

tinystories_path = str((current_dir / "../data/TinyStoriesV2-GPT4-valid.txt").resolve())
owt_path = str((current_dir / "../data/owt_valid.txt").resolve())

tiny_tokenizer = BPETokenizer.from_pretrained(
    vocab_path=tiny_vocab_path,
    merges_path=tiny_merges_path,
    )
owt_tokenizer = BPETokenizer.from_pretrained(
    vocab_path=owt_vocab_path,
    merges_path=owt_merges_path,
    )
```

```python
samples_tinystories, samples_owt = [], []
with open(tinystories_path, 'r') as f:
    possible_idx = list(range(0, 100))
    idx_selected = random.sample(possible_idx, 10)
    idx = 0
    while idx < 100:
        line = f.readline()
        if idx in idx_selected:
            samples_tinystories.append(line.strip())
        idx += 1
with open(owt_path, 'r') as f:
    possible_idx = list(range(0, 100))
    idx_selected = random.sample(possible_idx, 10)
    idx = 0
    while idx < 100:
        line = f.readline()
        if idx in idx_selected:
            samples_owt.append(line.strip())
        idx += 1
```
